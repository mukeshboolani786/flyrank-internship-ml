## 3. Human review + the no-go list

*What a person must check before acting. What should never be automated.*

### Human Review and the No-Go List

This section outlines the critical role of human review in the content action playbook and explicitly lists actions that should never be automated. Human oversight ensures ethical considerations, brand consistency, legal compliance, and strategic alignment that algorithms cannot fully address.

**Human Review Rules:**
*   **Mandatory Review for High-Impact Actions**: Any recommendation involving significant budget allocation, brand-sensitive messaging, or high-visibility content changes (e.g., homepage updates, major campaign launches) must undergo human review.
*   **Ethical Considerations**: Reviewers must assess recommendations for potential biases, fairness, and compliance with ethical guidelines, especially concerning audience targeting or content tone.
*   **Legal & Compliance Check**: All proposed content modifications or new content ideas must be reviewed for legal compliance (e.g., copyright, data privacy, accessibility standards) and adherence to industry regulations.
*   **Brand Voice & Tone**: Human editors are responsible for ensuring that all content adheres to the established brand voice, tone, and editorial guidelines, which can be difficult for algorithms to fully capture.
*   **Qualitative Assessment**: Reviewers should provide qualitative feedback on content quality, creativity, and resonance with the target audience, supplementing quantitative metrics.
*   **A/B Testing Approval**: Any proposed A/B tests generated by the system should be reviewed and approved by a human to ensure validity, ethical testing practices, and clear objectives.

**What Should NOT Be Automated (No-Go List):**
*   **Direct Content Generation**: While the system can recommend topics or keywords, the actual creation of written, visual, or audio content should not be fully automated to maintain quality, originality, and brand integrity.
*   **Sensitive Audience Targeting**: Automated targeting based on potentially discriminatory attributes or for sensitive topics without human oversight.
*   **Legal or Financial Advice**: Content that provides legal, financial, or medical advice must always be crafted and reviewed by qualified human experts.
*   **Crisis Communications**: Automated responses or content generation during a brand crisis or sensitive public relations situation is strictly prohibited.
*   **User-Generated Content (UGC) Moderation**: While AI can assist in flagging, final decisions on moderating or removing UGC should remain with human moderators to prevent false positives/negatives and ensure fairness.
*   **Major Strategic Shifts**: Decisions that fundamentally alter content strategy, target audience, or market positioning should be driven by human strategists, not solely by algorithmic recommendations.

### Monitoring / Retrain Triggers

This section defines the metrics and conditions that would indicate the content action playbook's recommendations are becoming stale or less effective, thereby triggering a need for model retraining or re-evaluation. Establishing clear triggers is essential for maintaining the playbook's relevance and performance over time.

**Key Monitoring Metrics and Triggers:**
*   **Performance Metrics Decay**: A consistent decline (e.g., 5-10% drop over 30 days) in key performance indicators (KPIs) directly impacted by the recommended actions, such as:
    *   **Click-Through Rate (CTR)**: For recommendations related to content visibility or promotion.
    *   **Conversion Rate**: For recommendations aimed at driving specific user actions (e.g., purchases, sign-ups).
    *   **Engagement Metrics**: (e.g., average time on page, bounce rate) for content quality recommendations.
    *   **SEO Rankings**: For actions targeting search engine visibility.
*   **Reason Code Efficacy**: A significant increase in the number of recommended actions that fail to produce the expected outcome (e.g., 'LowCTR_HighImpressions' actions still yielding low CTR after implementation).
*   **Data Drift Detection**: Monitoring changes in the distribution of input data features (e.g., user demographics, search query patterns, content consumption trends). Significant shifts (e.g., using statistical tests like KS-test or Earth Mover's Distance) could indicate the model is operating on outdated assumptions.
*   **Feedback Loop Inconsistencies**: An increase in human reviewers overriding or rejecting recommendations, suggesting a mismatch between model output and human expertise/intuition.
*   **External Events**: Major market shifts, product launches, competitive actions, or changes in platform algorithms (e.g., Google search updates) that could invalidate current recommendations.
*   **Model Confidence Scores**: A consistent drop in the model's prediction confidence for its recommendations, indicating growing uncertainty.
*   **Seasonal or Trend Changes**: The model might require re-evaluation or retraining during significant seasonal shifts or when a new major trend emerges that was not present in the training data.

**Retraining Protocol:**
When a trigger is activated, a predefined protocol should initiate, including:
1.  **Investigation**: Analyze the root cause of the performance decay or data drift.
2.  **Data Refresh**: Collect and preprocess new data reflecting current trends.
3.  **Model Retraining**: Retrain the model on the updated dataset.
4.  **Validation**: Re-validate the retrained model against hold-out data and A/B test if necessary.
5.  **Deployment**: Deploy the updated model and monitor its performance closely.
